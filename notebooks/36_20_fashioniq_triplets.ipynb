```python
import json
import os
import pandas as pd

def generate_triplet_csvs(input_dir, output_dir):
    """
    Parses FashionIQ JSON annotations and converts them into unified CSVs.
    """
    categories = ['dress', 'shirt', 'toptee']
    splits = ['train', 'val', 'test']

    # Ensure the output directory exists
    os.makedirs(output_dir, exist_ok=True)

    for split in splits:
        triplet_records = []

        for category in categories:
            # FashionIQ standard naming convention
            filename = f"cap.{category}.{split}.json"
            filepath = os.path.join(input_dir, filename)

            if not os.path.exists(filepath):
                print(f"Warning: File not found - {filepath}")
                continue

            with open(filepath, 'r', encoding='utf-8') as f:
                data = json.load(f)

            for item in data:
                # Extract the core triplet components
                candidate = item.get('candidate', '')
                target = item.get('target', '')

                # FashionIQ provides a list of 2 captions describing the visual change.
                # We combine them into a single coherent modifier string.
                captions = item.get('captions', [])
                modifier = " and ".join(captions)

                triplet_records.append({
                    'candidate': candidate,
                    'modifier': modifier,
                    'target': target,
                    'category': category # Keeping category for your Week 17 evaluation
                })

        # Save to CSV if records were found
        if triplet_records:
            df = pd.DataFrame(triplet_records)
            output_file = os.path.join(output_dir, f"fashioniq_triplets_{split}.csv")
            df.to_csv(output_file, index=False, encoding='utf-8')
            print(f"Thành công: Đã lưu {len(df)} triplets vào {output_file}")

if __name__ == "__main__":
    # Define your paths here
    INPUT_JSON_DIR = "/content/drive/MyDrive/GRADUATION_THESIS/captions"
    OUTPUT_CSV_DIR = "/content/drive/MyDrive/GRADUATION_THESIS/processed_data"

    print("Bắt đầu xử lý dữ liệu FashionIQ...")
    generate_triplet_csvs(INPUT_JSON_DIR, OUTPUT_CSV_DIR)
```
